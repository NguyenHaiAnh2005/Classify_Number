```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Đọc dữ liệu
train_data = np.load("data/raw/mnist_train.npz")
test_data = np.load("data/raw/mnist_test.npz")

# Xem các key bên trong file
print(train_data.files)
print(test_data.files)
```

```python
X_train = train_data['X']
y_train = train_data['y']
for i in range(10):
    plt.imshow(X_train[i], cmap="gray")
    plt.axis("off")
    plt.title(f"{y_train[i]}")
    plt.show()
```

```python
print(X_train.shape,y_train.shape,sep ='\n')
```

```python
for i in range(10):
    image = X_train[y_train == i].mean(axis = 0)
    plt.imshow(image, cmap="gray")
    plt.axis("off")
    plt.title(f"{i}")
    plt.show()
```

```python
y = pd.DataFrame(y_train)
```

```python
y.describe()
```

```python
data = y.value_counts()
print(data)
```

```python
for i in range(10):
   plt.bar(i,data[i])
for i in range(10):
    plt.text(i,data[i], str(data[i]),
      ha="center", va="bottom")
plt.title('Count label')
plt.xlabel('Label')
plt.ylabel('Count')
```
